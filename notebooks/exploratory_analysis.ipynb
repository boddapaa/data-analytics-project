# Sales Data Exploratory Analysis

This notebook performs exploratory data analysis on sales data and generates visualizations saved to the `/plots` folder.

## 1. Setup and Data Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Create plots directory if it doesn't exist
plots_dir = '../plots'
os.makedirs(plots_dir, exist_ok=True)

print('Setup complete!')
```

```python
# Load the data
df = pd.read_csv('../data/sample_sales_data.csv')
df['date'] = pd.to_datetime(df['date'])

print(f"Dataset shape: {df.shape}")
print(f"\nFirst few rows:")
df.head()
```

## 2. Data Overview

```python
# Display basic information
print("Dataset Info:")
df.info()

print("\n" + "="*50)
print("Summary Statistics:")
df.describe()
```

```python
# Check for missing values
print("Missing Values:")
print(df.isnull().sum())

print("\nUnique Values:")
for col in df.columns:
    print(f"{col}: {df[col].nunique()}")
```

## 3. Sales Analysis

```python
# Calculate key metrics
total_sales = df['sales'].sum()
total_quantity = df['quantity'].sum()
avg_sale = df['sales'].mean()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Average Sale: ${avg_sale:.2f}")
print(f"Number of Transactions: {len(df)}")
```

## 4. Visualizations (Saved to /plots)

```python
# Visualization 1: Sales by Category
plt.figure(figsize=(10, 6))
category_sales = df.groupby('category')['sales'].sum().sort_values(ascending=False)
category_sales.plot(kind='bar', color='steelblue')
plt.title('Total Sales by Category', fontsize=16, fontweight='bold')
plt.xlabel('Category', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# Save to plots folder
plt.savefig(os.path.join(plots_dir, 'notebook_sales_by_category.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"✓ Saved to {plots_dir}/notebook_sales_by_category.png")
```

```python
# Visualization 2: Sales by Region
plt.figure(figsize=(10, 8))
region_sales = df.groupby('region')['sales'].sum()
colors = plt.cm.Set3(range(len(region_sales)))
plt.pie(region_sales, labels=region_sales.index, autopct='%1.1f%%', 
        colors=colors, startangle=90)
plt.title('Sales Distribution by Region', fontsize=16, fontweight='bold')
plt.axis('equal')

# Save to plots folder
plt.savefig(os.path.join(plots_dir, 'notebook_sales_by_region.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"✓ Saved to {plots_dir}/notebook_sales_by_region.png")
```

```python
# Visualization 3: Daily Sales Trend
plt.figure(figsize=(14, 6))
daily_sales = df.groupby('date')['sales'].sum().sort_index()
plt.plot(daily_sales.index, daily_sales.values, marker='o', 
         linewidth=2, markersize=4, color='darkblue')
plt.title('Daily Sales Trend', fontsize=16, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save to plots folder
plt.savefig(os.path.join(plots_dir, 'notebook_daily_sales_trend.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"✓ Saved to {plots_dir}/notebook_daily_sales_trend.png")
```

```python
# Visualization 4: Top Products
plt.figure(figsize=(10, 8))
product_sales = df.groupby('product')['sales'].sum().sort_values(ascending=True)
product_sales.plot(kind='barh', color='coral')
plt.title('Total Sales by Product', fontsize=16, fontweight='bold')
plt.xlabel('Total Sales ($)', fontsize=12)
plt.ylabel('Product', fontsize=12)
plt.tight_layout()

# Save to plots folder
plt.savefig(os.path.join(plots_dir, 'notebook_top_products.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"✓ Saved to {plots_dir}/notebook_top_products.png")
```

```python
# Visualization 5: Category-Region Heatmap
plt.figure(figsize=(10, 6))
pivot_data = df.pivot_table(values='sales', index='category', 
                             columns='region', aggfunc='sum', fill_value=0)
sns.heatmap(pivot_data, annot=True, fmt='.0f', cmap='YlOrRd', 
            linewidths=0.5, cbar_kws={'label': 'Total Sales ($)'})
plt.title('Sales Heatmap: Category vs Region', fontsize=16, fontweight='bold')
plt.xlabel('Region', fontsize=12)
plt.ylabel('Category', fontsize=12)
plt.tight_layout()

# Save to plots folder
plt.savefig(os.path.join(plots_dir, 'notebook_category_region_heatmap.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"✓ Saved to {plots_dir}/notebook_category_region_heatmap.png")
```

## 5. Statistical Analysis

```python
# Sales statistics by category
print("Sales Statistics by Category:")
df.groupby('category')['sales'].agg(['count', 'sum', 'mean', 'std', 'min', 'max'])
```

```python
# Sales statistics by region
print("Sales Statistics by Region:")
df.groupby('region')['sales'].agg(['count', 'sum', 'mean', 'std', 'min', 'max'])
```

```python
# Product performance analysis
print("Product Performance Analysis:")
product_analysis = df.groupby('product').agg({
    'sales': ['sum', 'mean', 'count'],
    'quantity': 'sum'
}).round(2)
product_analysis.columns = ['Total Sales', 'Avg Sales', 'Transactions', 'Total Quantity']
product_analysis.sort_values('Total Sales', ascending=False)
```

## 6. Key Insights

```python
# Generate insights
print("KEY INSIGHTS:")
print("="*50)

# Best performing category
best_category = df.groupby('category')['sales'].sum().idxmax()
best_category_sales = df.groupby('category')['sales'].sum().max()
print(f"1. Best Performing Category: {best_category} (${best_category_sales:,.2f})")

# Best performing region
best_region = df.groupby('region')['sales'].sum().idxmax()
best_region_sales = df.groupby('region')['sales'].sum().max()
print(f"2. Best Performing Region: {best_region} (${best_region_sales:,.2f})")

# Top product
top_product = df.groupby('product')['sales'].sum().idxmax()
top_product_sales = df.groupby('product')['sales'].sum().max()
print(f"3. Top Product: {top_product} (${top_product_sales:,.2f})")

# Average transaction value
avg_transaction = df['sales'].mean()
print(f"4. Average Transaction Value: ${avg_transaction:.2f}")

# Date range
date_range = f"{df['date'].min().strftime('%Y-%m-%d')} to {df['date'].max().strftime('%Y-%m-%d')}"
print(f"5. Data Period: {date_range}")

print("="*50)
print("\n✓ Analysis complete! All visualizations saved to /plots folder.")
```
